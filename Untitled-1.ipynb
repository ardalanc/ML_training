```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from seaborn import scatterplot
from sklearn import linear_model, preprocessing
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import KFold
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import accuracy_score, r2_score
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
#%%
df = pd.read_csv(r"E:\dt_prj\housing.csv")
#%%
x = df.drop("median_house_value", axis=1)
y = df["median_house_value"]
#%%
df.head()
#%%
train_x , test_x , train_y , test_y = train_test_split(x,y,test_size=0.2, random_state=42 , shuffle=True)
#%%
train_x_num = np.array(train_x.drop("ocean_proximity", axis=1))
train_x_cat = np.array((train_x[["ocean_proximity"]]))
test_x_num = np.array((test_x.drop("ocean_proximity", axis=1)))
test_x_cat = np.array((test_x[["ocean_proximity"]]))
num_atributes = list(train_x.select_dtypes(include=[np.float64]).columns)
cat_atributes = ["ocean_proximity"]
#%%
num_pipline = Pipeline([
    ('imputer', SimpleImputer(strategy="median")),
    ('scaler', StandardScaler()),
    ('poly', PolynomialFeatures(degree=2, include_bias=False)),
])
cat_pipline = Pipeline([
    ("encode" , OneHotEncoder(sparse_output=False)),
])
#%%
preprocessing_stage = ColumnTransformer([
    ("num", num_pipline , num_atributes),
    ("cat", cat_pipline , cat_atributes),
])
#%%
full_pipline = Pipeline([
    ("preprocessing_stage" , preprocessing_stage),
    ("model" , LinearRegression()),
])
#%%
full_pipline.fit(train_x, train_y)
#%%
param_grids = {"preprocessing_stage__num__poly__degree":[2,3,4,5]}
grid_search = GridSearchCV(estimator=full_pipline,param_grid=param_grids,scoring="r2",cv=5,n_jobs=-1,verbose=2)
grid_search.fit(train_x, train_y)
#%%
print(f"best parametrs : {grid_search.best_params_}")
#%%
print(f"best score : {grid_search.best_score_*100}")
#%%
best_model = grid_search.best_estimator_
#%%
pred_test_y_new = best_model.predict(test_x)
#%%
```
